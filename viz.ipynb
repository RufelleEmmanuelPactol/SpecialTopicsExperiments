```python
import pandas as pd
import plotly.express as px

data = {
    'Date': ['2025-04-07', '2025-03-31', '2025-03-24', '2025-03-17'],
    'Users': [1983, 2708, 2708, 2640],
    'Revenue': [15079, 17019, 17341, 16393]
}

df = pd.DataFrame(data)
df['Date'] = pd.to_datetime(df['Date'])
df['RevenuePerUser'] = df['Revenue'] / df['Users']

fig = px.line(
    df,
    x='Date',
    y='RevenuePerUser',
    title='Revenue per User Over Time',
    markers=True,
    labels={'RevenuePerUser': 'Revenue per User'}
)
fig.update_yaxes(range=[3, df['RevenuePerUser'].max() * 1.1])
fig.write_image("revenue_per_user.png", width=800, height=600)
```
